## 1. What is TypeScript & Why Use It?

TypeScript is a superset of JavaScript that adds static types. It compiles to plain JavaScript but catches errors at compile-time rather than runtime.

**Key Benefits:**
- Catch bugs early (type errors before running code)
- Better IDE support (autocomplete, refactoring)
- Self-documenting code (types act as inline documentation)
- Easier refactoring (compiler ensures consistency)

**Simple Example:**

````typescript
// JavaScript (no type safety)
function add(a, b) {
  return a + b;
}

add(5, 3);        // 8
add('5', '3');    // '53' (oops, string concatenation)
add(5, '3');      // '53' (type coercion, hard to spot bugs)

// TypeScript (with types)
function addTS(a: number, b: number): number {
  return a + b;
}

addTS(5, 3);      // 8
addTS('5', 3);    // ❌ Error: Argument of type 'string' is not assignable to parameter of type 'number'
````

## 2. Basic Types and Type Annotations

TypeScript supports primitive types, arrays, tuples, enums, and more.

````typescript
// Primitive types
const str: string = 'hello';
const num: number = 42;
const bool: boolean = true;
const nothing: null = null;
const undef: undefined = undefined;
const any_val: any = 'anything goes';  // Avoid 'any' when possible

// Arrays
const nums: number[] = [1, 2, 3];
const strs: Array<string> = ['a', 'b'];  // Generic syntax

// Tuples (fixed-length arrays with specific types at each position)
const pair: [string, number] = ['age', 25];
const triple: [boolean, number, string] = [true, 42, 'done'];

// Union types (can be one of several types)
let id: number | string;
id = 123;        // ok
id = 'abc123';   // ok
// id = true;    // ❌ Error

// Type literals
let status: 'pending' | 'success' | 'error';
status = 'pending';
// status = 'pending2';  // ❌ Error

console.log(str, num, bool, nums, pair);
````

### Enums

Enums allow naming sets of numeric or string constants.

````typescript
// Numeric enum
enum Direction {
  Up = 1,
  Down = 2,
  Left = 3,
  Right = 4
}

let dir: Direction = Direction.Up;  // 1
console.log(dir);

// String enum
enum Status {
  Active = 'ACTIVE',
  Inactive = 'INACTIVE',
  Pending = 'PENDING'
}

let s: Status = Status.Active;
console.log(s);  // 'ACTIVE'
````

## 3. Interfaces and Object Shapes

Interfaces define contracts for object shape—what properties and methods an object must have.

````typescript
interface User {
  id: number;
  name: string;
  email: string;
  age?: number;  // Optional property
}

const user1: User = {
  id: 1,
  name: 'Alice',
  email: 'alice@example.com',
  age: 25
};

const user2: User = {
  id: 2,
  name: 'Bob',
  email: 'bob@example.com'
  // age is optional, so ok to omit
};

// ❌ Missing required properties
// const user3: User = { id: 3, name: 'Charlie' };

console.log(user1, user2);
````

### Extending Interfaces

Interfaces can extend other interfaces, adding more constraints.

````typescript
interface Admin extends User {
  role: 'admin' | 'mod' | 'user';
  permissions: string[];
}

const admin: Admin = {
  id: 10,
  name: 'Eve',
  email: 'eve@example.com',
  role: 'admin',
  permissions: ['delete', 'manage_users']
};

console.log(admin);
````

### Readonly Properties

Make properties immutable after initialization.

````typescript
interface Config {
  readonly apiUrl: string;
  readonly port: number;
  maxRetries?: number;
}

const cfg: Config = {
  apiUrl: 'https://api.example.com',
  port: 3000
};

console.log(cfg.apiUrl);
// cfg.apiUrl = 'https://other.com';  // ❌ Error: cannot assign to readonly
````

## 4. Functions: Typing Parameters and Returns

Annotate function parameters and return types for clarity and safety.

````typescript
// Function with typed parameters and return type
function greet(name: string, greeting: string = 'Hello'): string {
  return `${greeting}, ${name}!`;
}

console.log(greet('Alice'));
console.log(greet('Bob', 'Hi'));
// greet('Charlie', 123);  // ❌ Error: number not assignable to string

// Arrow function with types
const multiply = (a: number, b: number): number => a * b;
console.log(multiply(3, 4));

// Function returning void
function logMessage(msg: string): void {
  console.log(msg);
}

logMessage('This returns nothing');
````

### Optional and Default Parameters

````typescript
function describe(name: string, age?: number, city?: string): string {
  let desc = `Name: ${name}`;
  if (age) desc += `, Age: ${age}`;
  if (city) desc += `, City: ${city}`;
  return desc;
}

console.log(describe('Alice'));
console.log(describe('Bob', 30));
console.log(describe('Charlie', 25, 'NYC'));
````

### Function Types

````typescript
// Define a function type
type MathOp = (a: number, b: number) => number;

const add: MathOp = (a, b) => a + b;
const sub: MathOp = (a, b) => a - b;

console.log(add(5, 3));  // 8
console.log(sub(5, 3));  // 2

// Function that takes a function as parameter
function applyOp(x: number, y: number, op: MathOp): number {
  return op(x, y);
}

console.log(applyOp(10, 3, add));  // 13
console.log(applyOp(10, 3, sub));  // 7
````

## 5. Generics: Reusable Parameterized Types

Generics allow writing code that works with any type while maintaining type safety.

````typescript
// Generic function
function identity<T>(val: T): T {
  return val;
}

console.log(identity(5));        // 5
console.log(identity('hello'));  // 'hello'
console.log(identity([1, 2]));   // [1, 2]

// Generic with multiple type parameters
function pair<T, U>(first: T, second: U): [T, U] {
  return [first, second];
}

const p1 = pair(5, 'hello');
const p2 = pair('a', true);
console.log(p1, p2);
````

### Generic Constraints

Limit what types a generic can accept.

````typescript
// Generic type variable constrained to objects with 'length'
function getLength<T extends { length: number }>(val: T): number {
  return val.length;
}

console.log(getLength('hello'));      // 5
console.log(getLength([1, 2, 3]));    // 3
// getLength(123);  // ❌ Error: number has no 'length'

// Constrain to extend an interface
interface HasId {
  id: number;
}

function printId<T extends HasId>(obj: T): void {
  console.log(`ID: ${obj.id}`);
}

printId({ id: 1, name: 'Alice' });
// printId({ name: 'Bob' });  // ❌ Error: missing 'id'
````

### Generic Classes

Classes can also be parameterized.

````typescript
class Container<T> {
  private value: T;

  constructor(val: T) {
    this.value = val;
  }

  getValue(): T {
    return this.value;
  }

  setValue(val: T): void {
    this.value = val;
  }
}

const numContainer = new Container(42);
console.log(numContainer.getValue());  // 42

const strContainer = new Container('hello');
console.log(strContainer.getValue());   // 'hello'
````

## 6. Union and Intersection Types

Combine types to express complex relationships.

````typescript
// Union type: can be one of several types
type StringOrNumber = string | number;
type Success<T> = { status: 'ok'; data: T };
type Error = { status: 'error'; message: string };

type Result<T> = Success<T> | Error;

function handleResult<T>(res: Result<T>): void {
  if (res.status === 'ok') {
    console.log('Success:', res.data);
  } else {
    console.log('Error:', res.message);
  }
}

const res1: Result<string> = { status: 'ok', data: 'Hello' };
const res2: Result<string> = { status: 'error', message: 'Something failed' };

handleResult(res1);
handleResult(res2);
````

### Intersection Types

Combine multiple types into one (object must satisfy all).

````typescript
interface Named {
  name: string;
}

interface Aged {
  age: number;
}

type Person = Named & Aged;

const person: Person = {
  name: 'Alice',
  age: 30
};

console.log(person);

// Combining via intersection
type Admin = Person & { role: 'admin' };

const admin: Admin = {
  name: 'Eve',
  age: 35,
  role: 'admin'
};

console.log(admin);
````

## 7. Classes and Access Modifiers

TypeScript enhances classes with visibility modifiers and type safety.

````typescript
class Animal {
  name: string;
  age: number;

  constructor(name: string, age: number) {
    this.name = name;
    this.age = age;
  }

  describe(): string {
    return `${this.name} is ${this.age} years old`;
  }
}

const dog = new Animal('Rex', 5);
console.log(dog.describe());
````

### Access Modifiers: public, private, protected

````typescript
class BankAccount {
  public accountHolder: string;  // Accessible everywhere
  private balance: number;       // Only inside this class
  protected accountType: string; // Inside this class and subclasses

  constructor(holder: string, initialBalance: number, type: string) {
    this.accountHolder = holder;
    this.balance = initialBalance;
    this.accountType = type;
  }

  public getBalance(): number {
    return this.balance;
  }

  public deposit(amount: number): void {
    if (amount > 0) {
      this.balance += amount;
      console.log(`Deposited ${amount}`);
    }
  }

  private updateBalance(amt: number): void {
    this.balance += amt;
  }
}

const acc = new BankAccount('Alice', 1000, 'savings');
console.log(acc.accountHolder);      // ok: public
console.log(acc.getBalance());       // ok: public method
// console.log(acc.balance);         // ❌ Error: private
// acc.updateBalance(-500);          // ❌ Error: private
````

### Class Inheritance

````typescript
class Vehicle {
  protected brand: string;

  constructor(brand: string) {
    this.brand = brand;
  }

  describe(): string {
    return `This is a ${this.brand}`;
  }
}

class Car extends Vehicle {
  doors: number;

  constructor(brand: string, doors: number) {
    super(brand);  // Call parent constructor
    this.doors = doors;
  }

  describe(): string {
    return `${super.describe()} with ${this.doors} doors`;
  }
}

const myCar = new Car('Toyota', 4);
console.log(myCar.describe());
console.log(myCar.doors);
````

## 8. Type Guards and Type Narrowing

Refine types within conditional blocks for safer code.

````typescript
function processInput(input: string | number): void {
  if (typeof input === 'string') {
    console.log(`String length: ${input.length}`);
  } else {
    console.log(`Number value: ${input}`);
  }
}

processInput('hello');
processInput(42);

// instanceof guard
class Cat {
  meow(): void { console.log('Meow'); }
}

class Dog {
  bark(): void { console.log('Woof'); }
}

function makeSound(animal: Cat | Dog): void {
  if (animal instanceof Cat) {
    animal.meow();
  } else {
    animal.bark();
  }
}

const cat = new Cat();
const dog = new Dog();

makeSound(cat);
makeSound(dog);
````

### Custom Type Guards

````typescript
interface Fish {
  swim(): void;
}

interface Bird {
  fly(): void;
}

// Type guard function
function isFish(animal: Fish | Bird): animal is Fish {
  return 'swim' in animal;
}

function move(animal: Fish | Bird): void {
  if (isFish(animal)) {
    animal.swim();
  } else {
    animal.fly();
  }
}

const fish: Fish = { swim: () => console.log('Swimming') };
const bird: Bird = { fly: () => console.log('Flying') };

move(fish);
move(bird);
````

## 9. Advanced Patterns: Utility Types

TypeScript provides built-in utility types to transform and manipulate types.

````typescript
interface User {
  id: number;
  name: string;
  email: string;
}

// Partial<T> makes all properties optional
type PartialUser = Partial<User>;
const updates: PartialUser = { name: 'Alice' };  // ok, only name

// Required<T> makes all properties required
type RequiredUser = Required<User>;
// const incomplete: RequiredUser = { id: 1 };  // ❌ Error: missing props

// Record<K, T> creates object with specific keys and value type
type Roles = 'admin' | 'user' | 'guest';
const permissions: Record<Roles, string[]> = {
  admin: ['read', 'write', 'delete'],
  user: ['read', 'write'],
  guest: ['read']
};

console.log(permissions);

// Pick<T, K> selects subset of properties
type UserPreview = Pick<User, 'id' | 'name'>;
const preview: UserPreview = { id: 1, name: 'Alice' };

console.log(preview);
````

### Omit and Readonly

````typescript
interface Config {
  host: string;
  port: number;
  ssl: boolean;
  timeout: number;
}

// Omit<T, K> excludes specific properties
type SimpleConfig = Omit<Config, 'timeout'>;
const cfg: SimpleConfig = { host: 'localhost', port: 3000, ssl: true };

// Readonly<T> makes all properties readonly
type ReadonlyConfig = Readonly<Config>;
const roConfig: ReadonlyConfig = { host: 'localhost', port: 3000, ssl: true, timeout: 5000 };
// roConfig.port = 8000;  // ❌ Error: readonly

// ReturnType<T> extracts return type from function
type MyFunc = (x: number) => string;
type MyReturn = ReturnType<MyFunc>;  // string

console.log(cfg, roConfig);
````

## 10. From JavaScript to TypeScript

Practical tips for converting existing JavaScript to TypeScript.

````typescript
// Step 1: Start with loose types (any)
function processData(data: any): any {
  return data.length || data.value;
}

// Step 2: Gradually add types
interface DataInput {
  length?: number;
  value?: number;
}

function processDataV2(data: DataInput): number | undefined {
  return data.length ?? data.value;
}

// Step 3: Use strict inference where possible
const arr = [1, 2, 3];           // Inferred as number[]
const obj = { x: 10, y: 20 };    // Inferred as { x: number, y: number }
const str = 'hello';             // Inferred as string

console.log('Gradual typing example completed');
````

### Best Practices for Migration

- **Rename `.js` to `.ts`** and let the compiler find errors
- **Use interfaces** for object shapes shared across files
- **Leverage type inference** (don't over-annotate)
- **Use `unknown`** instead of `any` when uncertain
- **Enable strict mode** in `tsconfig.json` for best results
- **Start with external types** (e.g., `@types/node`) for libraries
- **Incrementally type** from leaf modules upward

### A Minimal tsconfig.json

```json
{
  "compilerOptions": {
    "target": "ES2020",
    "module": "commonjs",
    "strict": true,
    "esModuleInterop": true,
    "skipLibCheck": true,
    "forceConsistentCasingInFileNames": true,
    "outDir": "./dist"
  },
  "include": ["src/**/*"]
}
```
