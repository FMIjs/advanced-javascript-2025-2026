### Подбор и промяна на DOM

**Цел:** Създаване и промяна по HTML елементи в процеса на работа със страница. Ползва се за построяване на раздвижени потребителски интерфейси.

[MDN: Document Object Model (DOM)](https://developer.mozilla.org/en-US/docs/Web/API/Document_Object_Model)

```javascript
// Подбор и манипулация на DOM
const app = document.getElementById('app') || document.body;

const heading = document.createElement('h1');
heading.textContent = 'Манипулация на DOM';
app.appendChild(heading);

const list = document.createElement('ul');
['Елемент 1', 'Елемент 2', 'Елемент 3'].forEach(text => {
  const li = document.createElement('li');
  li.textContent = text;
  list.appendChild(li);
});
app.appendChild(list);

console.log('DOM създаден');
```

### Слушатели и делегиране на събитие

**Цел:** Отговор на взаимодействия на потребителя (щраквания, натискане на клавиши, изпращане на форми). Делегирането позволява прикачване на един слушател към родителски елемент, но обработка на събития за множество дъщерни елементи.

[MDN: addEventListener()](https://developer.mozilla.org/en-US/docs/Web/API/EventTarget/addEventListener) | [MDN: Event delegation](https://developer.mozilla.org/en-US/docs/Learn_web_development/JavaScript/Building_blocks/Events#event_delegation)

```javascript
// Основен слушател на събитие
const button = document.createElement('button');
button.textContent = 'Щракни ме';
button.style.padding = '10px 20px';
button.style.marginTop = '20px';

button.addEventListener('click', (e) => {
  alert('Бутонът е щракнат! Тип събитие: ' + e.type);
});

document.body.appendChild(button);

// Делегиране на събитие
const container = document.createElement('div');
container.style.marginTop = '20px';

const itemList = document.createElement('ul');
itemList.id = 'item-list';

['Ябълка', 'Банан', 'Череша'].forEach(fruit => {
  const li = document.createElement('li');
  li.textContent = fruit;
  li.style.cursor = 'pointer';
  li.style.padding = '5px';
  itemList.appendChild(li);
});

// Един слушател на родителя обработва всички щраквания на дъщерни елементи
itemList.addEventListener('click', (e) => {
  if (e.target.tagName === 'LI') {
    console.log('Щракнат:', e.target.textContent);
    e.target.style.backgroundColor = '#ffeb3b';
  }
});

container.appendChild(itemList);
document.body.appendChild(container);
```

### Fetch API

**Цел:** Изпращане на HTTP заявки към сървъра и работа с отдалечени API от браузъра. отложена оценка на резултата.

[MDN: Fetch API](https://developer.mozilla.org/en-US/docs/Web/API/Fetch_API)

```javascript
// Fetch API
const fetchBtn = document.createElement('button');
fetchBtn.textContent = 'Доставяне на данни';
fetchBtn.style.marginTop = '20px';
fetchBtn.style.padding = '10px 20px';

const resultDiv = document.createElement('div');
resultDiv.style.marginTop = '10px';
resultDiv.style.padding = '10px';
resultDiv.style.border = '1px solid #ddd';

fetchBtn.addEventListener('click', async () => {
  resultDiv.textContent = 'Зареждане...';
  try {
    const res = await fetch('https://jsonplaceholder.typicode.com/users/1');
    if (!res.ok) throw new Error(`HTTP ${res.status}`);
    const user = await res.json();
    resultDiv.textContent = `Име: ${user.name}\nEmail: ${user.email}\nТелефон: ${user.phone}`;
  } catch (err) {
    resultDiv.textContent = 'Грешка: ' + err.message;
  }
});

document.body.appendChild(fetchBtn);
document.body.appendChild(resultDiv);
```

### localStorage & sessionStorage

**Цел:** Съхраняване на малки количества данни на клиентската страна — localStorage оцелява при презареждане на страниците, sessionStorage се изчиства, когато разделът се затвори.

[MDN: Web Storage API](https://developer.mozilla.org/en-US/docs/Web/API/Web_Storage_API)

```javascript
// localStorage и sessionStorage
// Съхраняване в постоянното хранилище
localStorage.setItem('user', JSON.stringify({
  name: 'Иван',
  email: 'ivan@example.com',
  timestamp: new Date().toISOString()
}));

// Съхраняване във временното хранилище (изчистено при затваряне на раздел)
sessionStorage.setItem('sessionId', Math.random().toString(36).substring(7));

// Извличане
const savedUser = JSON.parse(localStorage.getItem('user'));
const sessionId = sessionStorage.getItem('sessionId');
console.log('Съхранен потребител:', savedUser);
console.log('ID на сесия:', sessionId);

// Показване на страницата
const storageDiv = document.createElement('div');
storageDiv.style.marginTop = '20px';
storageDiv.style.padding = '10px';
storageDiv.style.border = '1px solid #ccc';
storageDiv.textContent = `Демонстрация на хранилище\nlocalStorage: Потребител: ${savedUser.name}\nsessionStorage: Сесия: ${sessionId}`;
document.body.appendChild(storageDiv);
```

### Geolocation API

**Цел:** Достъп до географското местоположение на потребителя (след разрешение) за услуги, базирани на местоположение, и картография.

[MDN: Geolocation API](https://developer.mozilla.org/en-US/docs/Web/API/Geolocation_API)

```javascript
// Geolocation API
const geoDiv = document.createElement('div');
geoDiv.style.marginTop = '20px';
geoDiv.style.padding = '10px';
geoDiv.style.border = '1px solid #333';

const geoBtn = document.createElement('button');
geoBtn.textContent = 'Получи местоположение';
geoBtn.style.marginRight = '10px';

const geoResult = document.createElement('p');
geoResult.textContent = 'Щракни на бутона, за да получиш местоположението си';

geoBtn.addEventListener('click', () => {
  geoResult.textContent = 'Получаване на местоположение...';
  
  navigator.geolocation.getCurrentPosition(
    (pos) => {
      const { latitude, longitude, accuracy } = pos.coords;
      geoResult.textContent = `Местоположение: ${latitude.toFixed(4)}, ${longitude.toFixed(4)}\nТочност: ${accuracy.toFixed(0)}м`;
    },
    (err) => {
      geoResult.textContent = 'Грешка: ' + err.message;
    },
    { timeout: 10000 }
  );
});

geoDiv.appendChild(geoBtn);
geoDiv.appendChild(geoResult);
document.body.appendChild(geoDiv);
```

### Notifications API

**Цел:** Изпращане на известия за предупреждаване на потребителите, включително извън браузърния прозорец (с разрешение).

[MDN: Notifications API](https://developer.mozilla.org/en-US/docs/Web/API/Notifications_API)

```javascript
// Notifications API (Push Notifications)
const notifDiv = document.createElement('div');
notifDiv.style.marginTop = '20px';
notifDiv.style.padding = '10px';
notifDiv.style.border = '1px solid #333';

const notifBtn = document.createElement('button');
notifBtn.textContent = 'Поиск на разрешение за известия';
notifBtn.style.marginRight = '10px';

const sendBtn = document.createElement('button');
sendBtn.textContent = 'Изпрати известие';
sendBtn.disabled = true;

const notifStatus = document.createElement('p');
notifStatus.textContent = 'Статус на разрешенията: ' + Notification.permission;

notifBtn.addEventListener('click', async () => {
  try {
    const permission = await Notification.requestPermission();
    notifStatus.textContent = 'Статус на разрешенията: ' + permission;
    sendBtn.disabled = permission !== 'granted';
  } catch (err) {
    notifStatus.textContent = 'Грешка: ' + err.message;
  }
});

sendBtn.addEventListener('click', () => {
  if (Notification.permission === 'granted') {
    new Notification('Здравей!', {
      body: 'Това е известие от JavaScript',
      icon: 'data:image/svg+xml,%3Csvg xmlns=%22http://www.w3.org/2000/svg%22 width=%2264%22 height=%2264%22%3E%3Ccircle cx=%2232%22 cy=%2232%22 r=%2232%22 fill=%22%234CAF50%22/%3E%3C/svg%3E'
    });
  }
});

notifDiv.appendChild(notifBtn);
notifDiv.appendChild(sendBtn);
notifDiv.appendChild(notifStatus);
document.body.appendChild(notifDiv);
```

### History API

**Цел:** Промяна браузърната история и URL адреси без пълно преоразмеряване на страниците за приложения с една страница.

[MDN: History API](https://developer.mozilla.org/en-US/docs/Web/API/History_API)

```javascript
// History API
const historyDiv = document.createElement('div');
historyDiv.style.marginTop = '20px';
historyDiv.style.padding = '10px';
historyDiv.style.border = '1px solid #333';

const pushBtn = document.createElement('button');
pushBtn.textContent = 'Добавя състояние';
pushBtn.style.marginRight = '5px';

const backBtn = document.createElement('button');
backBtn.textContent = 'Върни се';

const historyStatus = document.createElement('p');
historyStatus.textContent = 'Текущо състояние: начално';

let stateCount = 0;

pushBtn.addEventListener('click', () => {
  stateCount++;
  history.pushState(
    { page: stateCount },
    'Страница ' + stateCount,
    '/page' + stateCount
  );
  historyStatus.textContent = 'Подадено състояние: ' + stateCount + ', URL: /page' + stateCount;
});

backBtn.addEventListener('click', () => {
  history.back();
  historyStatus.textContent = 'Върнати се в историята';
});

window.addEventListener('popstate', (e) => {
  const state = e.state ? e.state.page : 'начално';
  historyStatus.textContent = 'Върнато състояние: ' + state;
});

historyDiv.appendChild(pushBtn);
historyDiv.appendChild(backBtn);
historyDiv.appendChild(historyStatus);
document.body.appendChild(historyDiv);
```

### Clipboard API

**Цел:** Програмно писане и читане от буфера за обмяна за функционалност за копиране/поставяне.

[MDN: Clipboard API](https://developer.mozilla.org/en-US/docs/Web/API/Clipboard_API)

```javascript
// Clipboard API
const clipboardDiv = document.createElement('div');
clipboardDiv.style.marginTop = '20px';
clipboardDiv.style.padding = '10px';
clipboardDiv.style.border = '1px solid #333';

const copyBtn = document.createElement('button');
copyBtn.textContent = 'Копирай в буфера';
copyBtn.style.marginRight = '10px';

const pasteBtn = document.createElement('button');
pasteBtn.textContent = 'Постави от буфера';

const output = document.createElement('p');

copyBtn.addEventListener('click', async () => {
  const text = 'Здравей от JavaScript!';
  try {
    await navigator.clipboard.writeText(text);
    output.textContent = 'Копирано: ' + text;
  } catch (err) {
    output.textContent = 'Копирането не успя: ' + err.message;
  }
});

pasteBtn.addEventListener('click', async () => {
  try {
    const text = await navigator.clipboard.readText();
    output.textContent = 'Поставено: ' + text;
  } catch (err) {
    output.textContent = 'Поставянето не успя: ' + err.message;
  }
});

clipboardDiv.appendChild(copyBtn);
clipboardDiv.appendChild(pasteBtn);
clipboardDiv.appendChild(output);
document.body.appendChild(clipboardDiv);
```

### Storage Events

**Цел:** Слушане за промени в хранилището от други разделите, за да се синхронизират данни в същия произход.

[MDN: storage event](https://developer.mozilla.org/en-US/docs/Web/API/Window/storage_event)

```javascript
// Storage Events (синхронизиране между разделители)
const syncDiv = document.createElement('div');
syncDiv.style.marginTop = '20px';
syncDiv.style.padding = '10px';
syncDiv.style.border = '1px solid #333';

const setKeyBtn = document.createElement('button');
setKeyBtn.textContent = 'Задай ключ на хранилище';
setKeyBtn.style.marginRight = '10px';

const stat = document.createElement('p');
stat.textContent = 'Отвори тази страница в друг раздел и променяй хранилището, за да видиш събитията';

setKeyBtn.addEventListener('click', () => {
  localStorage.setItem('sharedData', JSON.stringify({
    message: 'Здравей от раздел ' + Date.now(),
    time: new Date().toLocaleTimeString()
  }));
  stat.textContent = 'Хранилище актуализирано в ' + new Date().toLocaleTimeString();
});

// Слушание за промени в хранилището от други разделители
window.addEventListener('storage', (e) => {
  if (e.key === 'sharedData') {
    const data = JSON.parse(e.newValue);
    stat.textContent = 'Получено от друг раздел: ' + data.message;
  }
});

syncDiv.appendChild(setKeyBtn);
syncDiv.appendChild(stat);
document.body.appendChild(syncDiv);
```

### IndexedDB

**Цел:** Съхраняване на големи количества структурирани данни на клиентската страна с пълни възможности на база данни (транзакции, индексиране, заявки).

[MDN: IndexedDB API](https://developer.mozilla.org/en-US/docs/Web/API/IndexedDB_API)

```javascript
// IndexedDB (Голямо хранилище на обекти)
const dbDiv = document.createElement('div');
dbDiv.style.marginTop = '20px';
dbDiv.style.padding = '10px';
dbDiv.style.border = '1px solid #333';

const dbBtn = document.createElement('button');
dbBtn.textContent = 'Инициализирай IndexedDB';
dbBtn.style.marginRight = '10px';

const addBtn = document.createElement('button');
addBtn.textContent = 'Добави потребител';
addBtn.disabled = true;

const dbStatus = document.createElement('p');
dbStatus.textContent = 'Щракни, за да инициализираш базата данни...';

let db;

dbBtn.addEventListener('click', () => {
  const req = indexedDB.open('courseDB', 1);
  
  req.onupgradeneeded = (e) => {
    db = e.target.result;
    const store = db.createObjectStore('users', { keyPath: 'id', autoIncrement: true });
    dbStatus.textContent = 'База данни надстроена, хранилище създадено';
  };
  
  req.onsuccess = (e) => {
    db = e.target.result;
    dbStatus.textContent = 'База данни отворена успешно';
    addBtn.disabled = false;
  };
  
  req.onerror = () => {
    dbStatus.textContent = 'Грешка при отваряне на базата данни';
  };
});

addBtn.addEventListener('click', () => {
  const tx = db.transaction('users', 'readwrite');
  const store = tx.objectStore('users');
  store.add({ name: 'Джейн', email: 'jane@example.com', joined: new Date() });
  
  tx.oncomplete = () => {
    dbStatus.textContent = 'Потребител добавен в IndexedDB';
  };
});

dbDiv.appendChild(dbBtn);
dbDiv.appendChild(addBtn);
dbDiv.appendChild(dbStatus);
document.body.appendChild(dbDiv);
```

### Navigator API

**Цел:** Разкриване на браузърни възможности, хардуерни ресурси (CPU ядра, памет на устройството) и наблюдаване на състояние на интернет връзката.

[MDN: Navigator](https://developer.mozilla.org/en-US/docs/Web/API/Navigator)

```javascript
// Navigator API (Браузърни възможности)
const navDiv = document.createElement('div');
navDiv.style.marginTop = '20px';
navDiv.style.padding = '10px';
navDiv.style.border = '1px solid #333';

const infoDiv = document.createElement('div');
infoDiv.style.fontFamily = 'monospace';
infoDiv.style.fontSize = '12px';
infoDiv.style.whiteSpace = 'pre-wrap';
infoDiv.style.wordWrap = 'break-word';

const info = {
  'User Agent': navigator.userAgent,
  'Език': navigator.language,
  'Онлайн': navigator.onLine ? 'Да' : 'Не',
  'Памет на устройството (GB)': navigator.deviceMemory || 'N/A',
  'Ядра': navigator.hardwareConcurrency || 'N/A',
  'Платформа': navigator.platform
};

let html = 'Информация за браузъра:\n';
Object.entries(info).forEach(([key, val]) => {
  html += key + ': ' + val + '\n';
});

infoDiv.textContent = html;

// Наблюдавай онлайн/офлайн
window.addEventListener('online', () => {
  infoDiv.textContent += '\n[Онлайн: Връзката е възстановена]';
});

window.addEventListener('offline', () => {
  infoDiv.textContent += '\n[Офлайн: Връзката е загубена]';
});

navDiv.appendChild(infoDiv);
document.body.appendChild(navDiv);
```

### Observer APIs

**Цел:** Наблюдаване на DOM елементи и промени в документите — пресичане за видимост, мутация за промени в DOM, преоразмеряване за размери на елементи.

[MDN: Intersection Observer API](https://developer.mozilla.org/en-US/docs/Web/API/Intersection_Observer_API) | [MDN: MutationObserver](https://developer.mozilla.org/en-US/docs/Web/API/MutationObserver) | [MDN: ResizeObserver](https://developer.mozilla.org/en-US/docs/Web/API/ResizeObserver)

```javascript
// Observer APIs (Пресичане, мутация, преоразмеряване)
const obsDiv = document.createElement('div');
obsDiv.style.marginTop = '20px';
obsDiv.style.padding = '10px';
obsDiv.style.border = '1px solid #333';

const log = document.createElement('div');
log.style.fontFamily = 'monospace';
log.style.fontSize = '12px';
log.style.maxHeight = '150px';
log.style.overflowY = 'auto';
log.style.border = '1px solid #ccc';
log.style.padding = '5px';
log.style.marginTop = '10px';
log.style.whiteSpace = 'pre-wrap';

const logMsg = (msg) => {
  log.textContent += msg + '\n';
  log.scrollTop = log.scrollHeight;
};

// Intersection Observer: откривай когато елемент попадне във видимост
const box = document.createElement('div');
box.style.width = '200px';
box.style.height = '100px';
box.style.backgroundColor = '#4CAF50';
box.style.color = 'white';
box.style.display = 'flex';
box.style.alignItems = 'center';
box.style.justifyContent = 'center';
box.textContent = 'Scroll или преоразмери, за да активираш';
box.id = 'observer-box';

const intersectionCb = (entries) => {
  entries.forEach(entry => {
    if (entry.isIntersecting) {
      logMsg('[Пресичане] Елемент в видимост');
    } else {
      logMsg('[Пресичане] Елемент извън видимост');
    }
  });
};

const intObserver = new IntersectionObserver(intersectionCb, { threshold: 0.5 });
intObserver.observe(box);

// Mutation Observer: откривай промени в DOM
const mutationCb = (mutations) => {
  mutations.forEach(m => {
    if (m.type === 'childList') {
      logMsg('[Мутация] Дъщерни елементи добавени/премахнати');
    } else if (m.type === 'attributes') {
      logMsg('[Мутация] Атрибут променен: ' + m.attributeName);
    }
  });
};

const mutObserver = new MutationObserver(mutationCb);
mutObserver.observe(box, {
  childList: true,
  attributes: true,
  subtree: true
});

// Resize Observer: откривай промени на размера
const resizeCb = (entries) => {
  entries.forEach(entry => {
    const { width, height } = entry.contentRect;
    logMsg('[Преоразмеряване] ' + width.toFixed(0) + 'x' + height.toFixed(0) + 'px');
  });
};

const resObserver = new ResizeObserver(resizeCb);
resObserver.observe(box);

const resizeBtn = document.createElement('button');
resizeBtn.textContent = 'Активирай преоразмеряване';
resizeBtn.style.marginRight = '10px';
resizeBtn.addEventListener('click', () => {
  box.style.width = (Math.random() * 300 + 100) + 'px';
});

obsDiv.appendChild(resizeBtn);
obsDiv.appendChild(box);
obsDiv.appendChild(log);
document.body.appendChild(obsDiv);
```

### Web Workers

**Цел:** Изпълнение на JavaScript код в отделен поток за изпълнение на тежки изчисления без блокиране на главния UI поток.

[MDN: Web Workers API](https://developer.mozilla.org/en-US/docs/Web/API/Web_Workers_API)

```javascript
// Web Workers - Изпълнение на код в отделен поток
const workerDiv = document.createElement('div');
workerDiv.style.marginTop = '20px';
workerDiv.style.padding = '10px';
workerDiv.style.border = '1px solid #333';

const inputBtn = document.createElement('button');
inputBtn.textContent = 'Изчисли 5^8 в Worker';
inputBtn.style.marginRight = '10px';

const resultDiv = document.createElement('p');
resultDiv.textContent = 'Щракни на бутона, за да изпълниш изчисление в фоновия поток';

inputBtn.addEventListener('click', () => {
  resultDiv.textContent = 'Worker изчислява...';
  
  // Създай worker от blob (встроен код)
  const workerCode = `
    self.onmessage = (e) => {
      const num = e.data;
      let result = 1;
      for (let i = 0; i < 8; i++) {
        result *= num;
      }
      self.postMessage({
        input: num,
        result: result,
        time: new Date().toLocaleTimeString()
      });
    }
  `;
  
  const blob = new Blob([workerCode], { type: 'application/javascript' });
  const worker = new Worker(URL.createObjectURL(blob));
  
  // Изпрати съобщение на worker
  worker.postMessage(5);
  
  // Получи резултат от worker
  worker.onmessage = (e) => {
    const { input, result, time } = e.data;
    resultDiv.textContent = `5^8 = ${result} (изчислено в ${time})`;
  };
});

workerDiv.appendChild(inputBtn);
workerDiv.appendChild(resultDiv);
document.body.appendChild(workerDiv);
```

### Модел на обвивка за хранилище

**Цел:** Създаване на преизползваем утилитарен модел за управление на localStorage с поддръжка на абонамент за синхронизиране между разделите.

[MDN: Web Storage API](https://developer.mozilla.org/en-US/docs/Web/API/Web_Storage_API)

```javascript
// Модел на обвивка за хранилище (преизползваем утилитарен модел)
const store = {
  set: (key, val) => {
    localStorage.setItem(key, JSON.stringify(val));
  },

  get: (key, def = null) => {
    const val = localStorage.getItem(key);
    return val ? JSON.parse(val) : def;
  },

  remove: (key) => {
    localStorage.removeItem(key);
  },

  clear: () => {
    localStorage.clear();
  },

  // Абониране за промени (от други разделители)
  watch: (callback) => {
    window.addEventListener('storage', callback);
    // Върни функция за отписване
    return () => window.removeEventListener('storage', callback);
  }
};

// Пример за използване
const wrapperDiv = document.createElement('div');
wrapperDiv.style.marginTop = '20px';
wrapperDiv.style.padding = '10px';
wrapperDiv.style.border = '1px solid #333';

const storeUserBtn = document.createElement('button');
storeUserBtn.textContent = 'Съхрани потребител';
storeUserBtn.style.marginRight = '10px';

const retrieveUserBtn = document.createElement('button');
retrieveUserBtn.textContent = 'Извлекай потребител';

const wrapperStatus = document.createElement('p');
wrapperStatus.textContent = 'Използвай обвивка, за да управляваш localStorage';

storeUserBtn.addEventListener('click', () => {
  store.set('currentUser', {
    id: 1,
    name: 'Алис',
    email: 'alice@example.com',
    theme: 'dark'
  });
  wrapperStatus.textContent = 'Потребител съхранен чрез обвивка';
});

retrieveUserBtn.addEventListener('click', () => {
  const user = store.get('currentUser');
  if (user) {
    wrapperStatus.textContent = 'Извлечен: ' + user.name + ' (' + user.email + ')';
  } else {
    wrapperStatus.textContent = 'Няма съхранен потребител';
  }
});

// Слушание на промени в хранилището
const unsubscribe = store.watch((e) => {
  if (e.key === 'currentUser') {
    wrapperStatus.textContent = 'Потребител променен от друг раздел';
  }
});

wrapperDiv.appendChild(storeUserBtn);
wrapperDiv.appendChild(retrieveUserBtn);
wrapperDiv.appendChild(wrapperStatus);
document.body.appendChild(wrapperDiv);
```
